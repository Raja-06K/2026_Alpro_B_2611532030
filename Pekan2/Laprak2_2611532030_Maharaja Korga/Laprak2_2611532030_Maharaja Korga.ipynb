## 2.1 Dasar Teori
1. **Variabel dan Aturan Penamaan**, Variabel adalah lokasi penyimpanan di memori yang diberi nama dan digunakan untuk menyimpan data. Pada Python, variabel dibuat secara otomatis saat nilai diberikan (assignment). Nama variabel harus bersifat deskriptif, diawali huruf atau underscore, dan tidak boleh menggunakan kata kunci (keyword) bawaan Python.

2. **Tipe Data Numerik (Numeric Types)**, Python mendukung beberapa jenis tipe data numerik:
    - Integer (int): Bilangan bulat positif atau negatif tanpa desimal (contoh: 100).
    - Float (float): Bilangan desimal/real (contoh: 10.2345).
    - Complex (complex): Bilangan kompleks yang terdiri dari bagian riil dan imajiner (contoh: 100+3j).

3. **Tipe Data String (str)**, Tipe data string digunakan untuk menyimpan sekumpulan karakter berupa teks. String dapat dideklarasikan menggunakan tanda petik tunggal ('...'), petik ganda ("..."), atau petik tiga ("""...""") untuk teks multibarise/paragraf.

4. **Tipe Data Boolean (bool)**, Tipe data boolean hanya memiliki dua nilai kebenaran, yaitu True (benar) atau False (salah). Tipe data ini umumnya digunakan untuk evaluasi kondisi dan alur kontrol percabangan.

5. **Fungsi Masukan (input) dan Konversi Tipe Data (Casting)**, Fungsi input() digunakan untuk menerima masukan dari pengguna secara interaktif. Hasil kembalian dari input() selalu bertipe string. Untuk melakukan operasi aritmatika, masukan harus dikonversi (type casting) menggunakan fungsi seperti int(), float(), atau sejenisnya.

6. **Konstanta (Constants)**, Konstanta adalah nilai yang bersifat tetap dan tidak boleh diubah sepanjang eksekusi program. Pada Python, batasan konstanta secara eksplisit dapat diindikasikan menggunakan penanda tipe Final dari modul typing.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611532030"
Subbab ini menjelaskan konsep tipe data boolean yang hanya memiliki dua nilai logika, yaitu True (benar) dan False (salah). Program ini memperlihatkan bagaimana logika boolean digunakan untuk mengevaluasi suatu kondisi ekspresi relasional (seperti perbandingan nilai) serta penggunaannya dalam pengondisian keputusan menggunakan pernyataan if dan operator and.

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data boolean
is_lulus_2030 = True
is_cumlaude_2030 = True

#Menggunakan Booloean
nilai_2030 = 85
batas_lulus_2030 = 75

#Menentukan nilai Boolean dari kondisi
status_kelulusan_2030 = nilai_2030 >= batas_lulus_2030 #Hasilnya akan True

print("=== Check Kelulusan===")
print("Nilai:", nilai_2030)
print("Apakah Lulus?:", status_kelulusan_2030)
if is_lulus_2030 and is_cumlaude_2030:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

### 2.2.2 Program "Input_2611532030"
Subbab ini membahas penggunaan fungsi bawaan input() untuk membuat program yang bersifat interaktif dengan pengguna. Selain mengambil masukan dari papan ketik (keyboard), program ini juga memperlihatkan proses konversi tipe data (type casting) dari string menjadi integer (int()) agar nilai yang dimasukkan dapat dihitung secara matematis, seperti pada rumus luas persegi panjang.  

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar prsegi panjang
# Nilai dimasukkan akan dikonversei menjadi tipe data integer
panjang_2030 = input('Masukkan nilai panjang : ')
lebar_2030 = input('Masukkan lebar : ')
luas_2030 = int(panjang_2030) * int(lebar_2030)
print("Luas =", luas_2030)
```

### 2.2.3 Program "Konstanta_2611532030"
Subbab ini membahas cara mendeklarasikan nilai variabel yang bersifat tetap atau tidak boleh diubah (konstanta) menggunakan modul typing.Final. Program ini mengimplementasikan konstanta nilai $\pi$ (PI) untuk menghitung luas lingkaran, dengan mengambil input nilai jari-jari dari pengguna serta menampilkan hasilnya menggunakan teknik format string %.2f.

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2030 = float(input('Masukkan nilai jari-jari: '))
luas_2030 = PI * jari_2030 * jari_2030
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2030, luas_2030))
```

### 2.2.4 Program "Numerik_2611532030"
Subbab ini membahas penggunaan tipe data numerik dalam Python, yang mencakup bilangan bulat (int), bilangan desimal/pecahan (float), serta bilangan imajiner (complex). Program ini bertujuan untuk menunjukkan cara pendeklarasian variabel bernilai angka serta memeriksa tipe data dari masing-masing variabel secara otomatis menggunakan fungsi bawaan type().

```python
# Buat file dengan nama Numerik_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
# Create a variable with numeric type value.
a_2030=100
print("The type of variable having value", a_2030, " is ", type(a_2030))

#create a variable with float value
b_2030=10.2345
print("The type of variable having value", b_2030, " is ", type(b_2030))

#create a variable with complex value 
c_2030=100+3j
print("The type of variable having value", c_2030, " is ", type(c_2030))
```

### 2.2.5 Program "String_2611532030"
Subbab ini berfokus pada pengolahan data teks atau karakter menggunakan tipe data string. Program ini mendemonstrasikan penulisan string menggunakan tanda petik tunggal ('), petik ganda ("), serta triple quotes (""") yang digunakan untuk mengelola teks berformat banyak baris (multiline) seperti alamat.  

```python
nama_2030 = "Maharaja_Korga"
jenis_kelamin_2030 = 'L'
alamat_2030 = """
    Lubuk Lintah
    kec.Kuranji
    kota Padang
"""
agama_2030 = "Islam"
print("Nama saya adalah:", nama_2030)
print("Jenis kelamin saya adalah:", jenis_kelamin_2030)
print("Alamat saya adalah:", alamat_2030)
print("agama saya adalah:", agama_2030)
```

**Penjelasan :**
1. Tipe Data Boolean: Hanya memiliki dua nilai logika, yaitu True (benar) atau False (salah). Umumnya digunakan untuk penentuan kondisi logis dalam kontrol keputusan.
2. Fungsi Input (input()): Fungsi bawaan Python untuk mengambil masukan teks dari pengguna. Mengingat hasil kembalian input() berupa tipe string, maka konversi tipe data (seperti int() atau float()) diperlukan jika input hendak diolah secara matematis.
3. Konstanta: Variabel yang nilainya bersifat tetap dan tidak boleh diubah setelah dideklarasikan. Di Python, variabel konstanta dapat dipertegas dengan modul typing.Final.
4. Tipe Data Numerik: Digunakan untuk menyimpan nilai berupa angka. Dalam Python terdapat beberapa jenis numerik, yaitu int (bilangan bulat), float (bilangan desimal/pecahan), dan complex (bilangan imajiner).
5. Tipe Data String: Digunakan untuk menyimpan sekumpulan karakter atau teks. String dapat diapit oleh tanda petik tunggal ('), petik ganda ("), maupun triple quotes (""") untuk teks multiline (berbaris-baris).
