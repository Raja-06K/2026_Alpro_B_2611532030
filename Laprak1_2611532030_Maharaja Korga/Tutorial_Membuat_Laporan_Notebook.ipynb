# Tutorial: Membuat Laporan Praktikum dengan Jupyter Notebook

Panduan ini menjelaskan cara memakai file **Template_LaporanPraktikum_StrukturData.ipynb** untuk membuat laporan praktikum, menggantikan format lama yang berupa file Word (.docx).

## 1. Persiapan Awal

1. Buat satu folder khusus untuk laporan, misalnya:
```
Laprak1_NIM_NamaMahasiswa/
├── Laporan.ipynb
└── images/
```
2. Salin file `Template_LaporanPraktikum_StrukturData.ipynb` ke folder tersebut, lalu rename sesuai format penamaan yang diminta (contoh: `Laprak1_2511531001_NamaMahasiswa.ipynb`).
3. Buat folder `images/` di dalam folder yang sama — untuk menyimpan logo, screenshot output, dan gambar soal.
4. Buka file dengan Jupyter Notebook, JupyterLab, VSCode (ekstensi Jupyter), atau Google Colab.

## 2. Mengisi Bagian Cover dan Kata Pengantar

- Cover dan Kata Pengantar sudah disediakan sebagai **markdown cell** paling atas pada template.
- Ganti teks placeholder (judul modul, nama, NIM, dosen, asisten) sesuai data masing-masing.
- Untuk logo kampus: taruh file logo di `images/logo_unand.png`, path di template sudah menyesuaikan otomatis.

## 3. Mengisi BAB I – BAB III

- Setiap BAB sudah dipisah sebagai markdown cell tersendiri (Pendahuluan, Pembahasan, Kesimpulan, Daftar Pustaka).
- Tulis ulang isi teks sesuai topik praktikum minggu tersebut.
- Gunakan heading (`#`, `##`, `###`) agar struktur tetap konsisten.

## 4. Menuliskan Kode Program

Ada dua kasus:

**a. Jika kode ditulis dalam Python**  
Tempel langsung ke *code cell*, lalu jalankan (`Shift + Enter`). Contoh di bawah ini:

````python
# Contoh: kode Python dijalankan langsung, output otomatis tersimpan di bawah cell
nilai = 10
print("Nilai adalah:", nilai)
````

**b. Jika kode ditulis dalam bahasa lain (Java, C, dsb — seperti pada contoh ADT)**

Notebook secara default menjalankan Python, jadi ada dua pilihan:

- **Opsi 1 (paling umum dipakai untuk laporan):** tempel kode sebagai teks di *code cell* Python hanya untuk ditampilkan rapi (tidak perlu di-run), lalu tempel hasil run dari IDE aslinya (Eclipse/NetBeans/VSCode) sebagai gambar di bawahnya.
- **Opsi 2 (jika ingin benar-benar dieksekusi di notebook):** gunakan magic command `%%writefile` untuk menulis kode ke file, lalu jalankan lewat perintah shell. Contoh untuk Java ada di cell berikutnya.

````python
%%writefile Contoh.java
public class Contoh {
    public static void main(String[] args) {
        System.out.println("Halo dari Java!");
    }
}
````

````python
!javac Contoh.java
!java Contoh
````

Output dari `!java` akan otomatis muncul di bawah cell, sama seperti output kode Python biasa — jadi tidak perlu screenshot manual.

## 5. Menambahkan Screenshot (Output Program, Soal Tugas, dsb)

**Cara A — dari file gambar di folder `images/`:**
```html
<div align="center">
<img src="images/output1.png" width="400"><br>
<em>Gambar 2.1 : Output Program</em>
</div>
```

**Cara B — copy-paste langsung ke markdown cell:**
1. Klik dua kali cell markdown untuk masuk mode edit.
2. Tempel gambar (Ctrl+V). Jupyter otomatis membuat baris ``.
3. Ubah jadi tag HTML supaya bisa diatur ukurannya:
```html

```
4. Jalankan cell (Shift+Enter) untuk melihat hasilnya.

> Tips: beri nama file gambar yang jelas (`output1.png`, `soal_tugas.png`, dll) dan urutkan penomoran gambar (Gambar 2.1, 2.2, dst) sesuai urutan kemunculan di laporan.

## 6. Mengisi Bagian Tugas Praktikum

- Bagian ini sudah disediakan strukturnya di template: Soal → Kode Jawaban (bisa lebih dari satu bagian, misal a. ADT dan b. Driver) → Output → Penjelasan.
- Soal tugas biasanya berupa gambar/screenshot dari modul — masukkan dengan cara yang sama seperti langkah 5.

## 7. Merapikan Sebelum Dikumpulkan

1. Jalankan seluruh notebook dari awal secara berurutan:  
   `Kernel/Run menu → Restart Kernel and Run All Cells`  
   Ini memastikan semua output konsisten dan tidak ada cell yang lupa dijalankan.
2. Cek semua gambar tampil dengan benar (tidak ada ikon gambar rusak/broken image).
3. Simpan notebook (`Ctrl+S`).

## 8. Format Pengumpulan

Tergantung permintaan dosen/format praktikum:

- **Kumpulkan file `.ipynb` langsung** — paling praktis, terutama jika gambar ditempel lewat cara *attachment* (Cara B) karena semua sudah menyatu dalam satu file.
- **Export ke PDF** (jika masih diminta satu file PDF seperti format lama):
  - JupyterLab: `File → Save and Export Notebook As... → PDF`
  - VSCode: klik ikon export di toolbar notebook, pilih PDF/HTML
  - Jika export ke PDF gagal karena butuh LaTeX, gunakan opsi **Export as HTML**, lalu buka HTML tersebut di browser dan `Print → Save as PDF`.

## 9. Penamaan File

Ikuti konvensi yang sama seperti laporan sebelumnya, contoh:
```
LaprakX_NIM_NamaMahasiswa.ipynb
```
di mana `X` adalah nomor pertemuan/pekan praktikum.
