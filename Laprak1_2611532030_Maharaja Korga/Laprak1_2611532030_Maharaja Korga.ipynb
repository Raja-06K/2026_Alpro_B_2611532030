## 2.2 Kode Program

### 2.2.1 Program "program1_2611532030"
Berikut adalah implementasi kode program pertama menggunakan sintaks dasar Python.

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\\\")
print("Program ini" + " dibuat oleh a\n" + "Maharaja Korga_2611532030.")
```

**Penjelasan :**
1. Tanda # (Hash / tanda pagar): digunakan untuk membuat komentar satu baris.
Komentar tidak dijalankan oleh Python, hanya sebagai catatan untuk programmer
2. Tiga tanda kutip """ (Triple quotation mark): dapat digunakan untuk menulis teks beberapa baris. Pada kode ini digunakan sebagai catatan.
3. Print(): berfungsi untuk menampilkan teks atau nilai ke layar.
4. Tanda () (Parentheses / tanda kurung): Menentukan isi/argumen fungsi, misalnya print()
4. Karakter \ (Backslash): Dalam Python, \ digunakan sebagai bagian dari escape sequence. Jika ingin menampilkan karakter \, biasanya ditulis sebagai: \\ (Double backslash)
5. Tanda + (Plus / operator penjumlahan): digunakan untuk menggabungkan beberapa string (teks).
6. Tanda \n (Newline): yaitu membuat pindah ke baris baru.
7. Tanda " (Double quotation mark): Menandai teks/string
