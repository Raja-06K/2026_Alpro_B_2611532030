## 2.1 Dasar Teori
Berikut adalah konsep dan teori dasar yang mendasari pelaksanaan praktikum perulangan dan nested loop dalam bahasa pemrograman Python:

1. Perulangan `for` pada Python
Perulangan `for` digunakan untuk melakukan iterasi terhadap urutan objek seperti deret angka, list, atau string. Pada Python, `for` bekerja secara terhitung (counted loop) dengan mengambil satu per satu elemen dari urutan yang diberikan.

2. Fungsi `range()`
Fungsi `range()` menghasilkan deret matematika berdasarkan parameter yang diberikan: `range(stop)`, `range(start, stop)`, atau `range(start, stop, step)`. Rentang nilai bersifat inklusif untuk `start` dan eksklusif untuk `stop` (berhenti tepat sebelum `stop`).

3. Perulangan Bersarang (Nested Loop)
Nested loop adalah struktur di mana sebuah perulangan berada di dalam blok perulangan lainnya. Loop luar (outer loop) mengontrol iterasi tingkat atas (seperti baris), sedangkan loop dalam (inner loop) menyelesaikan seluruh iterasinya pada setiap satu iterasi loop luar (seperti kolom).

4. Parameter `end` pada Fungsi `print()`
Secara bawaan (default), fungsi `print()` menambahkan karakter newline (`\\n`) di akhir output. Dengan mengubah parameter `end=""` atau `end=" "`, cetakan karakter berikutnya akan tetap berada pada baris yang sama tanpa berpindah baris.

5. Deret Palindromik dan Simetri Visual
Pola palindromik adalah deret susunan nilai yang simetris dari depan dan belakang. Dalam bentuk geometri dua dimensi, simetri ini dicapai dengan menggabungkan loop hitung mundur (reduksi) dan hitung maju (ekspansi) di sekitar titik pusat (singularity).

## 2.2 Kode Program

### 2.2.1 Program "Perulangan_for1_2611532030"

```python
# Buat file dengan nama perulangan_for1_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2030 = int(input("Inputkan angka: "))

for i_2030 in range(ulang_2030):
    print(f"Perulangan ke-{i_2030}")
```

**Penjelasan :**
1. Kode menerima masukan angka dari pengguna yang dikonversi ke tipe integer ke dalam variabel `ulang_2030`.\n",
2. Perulangan `for` menjalankan fungsi range(ulang_2030) yang menghasilkan deret angka mulai dari 0 hingga `ulang_2030 - 1`.
3. Pada setiap iterasi, variabel `i_2030` dicetak menggunakan formatted string untuk menampilkan nomor iterasi.

### 2.2.2 Program "Perulangan_for2_2611532030"

```python
# Buat file dengan nama perulangan_for2_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2030 = int(input("Inputkan angka: "))
print("Perulangan ke-0 sampai ke-", ulang_2030-1)
for i_2030 in range(ulang_2030):
    print(i_2030, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_2030)
for i_2030 in range(1, ulang_2030 + 1):
    print(i_2030, end=" ")
```

**Penjelasan :**
1. Program ini menampilkan dua variasi rentang `range()`.
2. Perulangan pertama menggunakan `range(ulang_2030)` untuk mencetak deret horizontal mulai dari indeks 0 hingga `N - 1` dengan parameter `end=\"\"`.
3. Perulangan kedua menggunakan `range(1, ulang_2030 + 1)` untuk mencetak deret dari angka 1 hingga `N` secara berurutan

### 2.2.3 Program "Perulangan_for3_2611532030"

```python
# Buat file dengan nama perulangan_for3_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2030 = int(input("Masukkan jumlah perulangan: "))

jumlah_2030 = 0
for i_2030 in range(1, ulang_2030 + 1):
    print(i_2030, end=" ")
    jumlah_2030 = jumlah_2030 + i_2030

    if i_2030 < ulang_2030:
        print("+", end=" ")
    else:
        print("= ", jumlah_2030, end=" ")
print()
print("Jumlah =", jumlah_2030)
```

**Penjelasan :**
1. Program menghitung akumulasi penjumlahan deret bilangan 1 sampai `N` menggunakan variabel `jumlah_2030`.
2. Struktur kondisi `if-else` di dalam loop menambahkan operator `+` jika angka belum mencapai batas, dan menampilkan tanda `=` beserta total hasil akumulasi pada akhir iterasi.

### 2.2.4 Program "Jumlah_genap_2611532030"

```python
# Buat file dengan nama jumlah_genap_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2030 = int(input("Masukkan nilai batas: "))

jumlah_2030 = 0
for i_2030 in range(1, ulang_2030 + 1):
    if i_2030 % 2 == 0:
        print(i_2030, end=" ")
        jumlah_2030 = jumlah_2030 + i_2030

        if i_2030 < ulang_2030:
            print("+", end=" ")
        else:
            print("= ", jumlah_2030, end=" ")
print()
print("Jumlah =", jumlah_2030)
```

**Penjelasan :**
1. Program melakukan pemfilteran angka genap menggunakan kondisi `i_2030 % 2 == 0` dari deret 1 hingga `N`.
2. Angka genap yang memenuhi syarat dijumlahkan ke dalam akumulator `jumlah_2030` dan dicetak dalam format deret penjumlahan.

### 2.2.5 Program "Nested_for1_2611532030"

```python
# Buat file dengan nama nested_for1_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2030 = int(input("Masukkan nilai batas: "))
for line_2030 in range(1, batas_2030 + 1):
    for j_2030 in range(1, (-1 * line_2030 + batas_2030) + 1):
        print(".", end="")
    print(line_2030)
```

**Penjelasan :**
1. Menggunakan nested loop untuk menyusun pola titik (`.`) dan nomor baris.
2. Loop luar `line_2030` menentukan baris aktif, sedangkan loop dalam `j_2030` mencetak jumlah titik berurutan menurun sesuai persamaan `(-1 * line_2030 + batas_2030) + 1` sebelum nomor baris dicetak

### 2.2.6 Program "Nested_for2_2611532030"

```python
# Buat file dengan nama nested_for2_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2030 = int(input("Masukkan nilai batas: "))
for i_2030 in range(1, batas_2030 + 1):
    for j_2030 in range(1, batas_2030 + 1):
        print("*", end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
1. Membentuk pola persegi bintang berukuran `batas_2030 x batas_2030`.
2. Loop dalam mencetak karakter `*`  secara horizontal sebanyak `batas_2030` kali, lalu `print()` kosong digunakan untuk berpindah ke baris baru.

### 2.2.7 Program "Nested_for3_2611532030"

```python
# Buat file dengan nama nested_for3_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2030 = int(input("Masukkan nilai batas: "))
for i_2030 in range(batas_2030+1):
    for j_2030 in range(batas_2030+1):
        print(i_2030 + j_2030, end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
1. Membuat matrik penjumlahan indeks `i_2030` dan `j_2030` mulai dari indeks 0 hingga `batas_2030`.
2. Setiap sel matrik menampilkan hasil penjumlahan variabel baris dan kolom secara terstruktur.

### 2.2.8 Program "Nested_for4_2611532030"

```python
# Buat file dengan nama nested_for4_nim.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit terakhir nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2030 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2030 % 2 != 0:
    print("Tinggi pola harus bilangan genap!")
else:
    a_2030 = tinggi_2030
    c_2030 = a_2030
    lebar_2030 = (2 * tinggi_2030) - 2

    for i_2030 in range(1, tinggi_2030 + 1):
        b_2030 = c_2030 + 1

        for j_2030 in range(1, lebar_2030 + 1):

            # Baris atas dan bawah
            if i_2030 == 1 or i_2030 == tinggi_2030:
                if j_2030 == 1 or j_2030 == lebar_2030:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_2030 == 1 or j_2030 == lebar_2030:
                    print("|", end="")
                else:
                    if j_2030 == c_2030:
                        print("<", end="")
                    elif j_2030 == b_2030:
                        print(">", end="")
                    elif j_2030 == (lebar_2030 - c_2030):
                        print("<", end="")
                    elif j_2030 == (lebar_2030 - c_2030 + 1):
                        print(">", end="")
                    elif j_2030 > b_2030 and j_2030 < (lebar_2030 - c_2030 + 1):
                        print(".", end="")
                    else:
                        print(" ", end="")
        
        print()

        # Logika asli java
        a_2030 -= 2

        if a_2030 <= 0:
            c_2030 = (-a_2030) + 2
        else:
            c_2030 = a_2030
```

**Penjelasan :**
1. Program ini membentuk pola jam pasir berbingkai kompleks menggunakan koordinat logika per baris dan kolom.
2. Memeriksa validasi input agar bernilai genap, lalu menghitung koordinat batas `<` dan `>` serta pengisian titik `.` atau spasi di dalamnya.

### Program "Latihan_pekan5_2611532030"

```python
# Tugas5 Buatlah program untuk menampilkan output segitiga bintang menggunakan perulangan for
tinggi_2030 = int(input("Masukkan tinggi segitiga: "))

for i_2030 in range(1, tinggi_2030 + 1):
    print(" " * (tinggi_2030 - i_2030), end="")

    for j_2030 in range(i_2030):
        print("*", end=" ")
    print()
```

**Penjelasan :**

## Tugas Praktikum

**Soal :**

Studi Kasus: Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai pembatas berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang merepresentasikan ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).

**Kode Jawaban :**

a. Kode Program "tugas5_2611532030.py"

```python
# ==============================================================================
# Program Jam Pasir Kristal Palindromik Berbingkai
# ==============================================================================

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

# Input nilai N dari pengguna
n_1234 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ------------------------------------------------------------------------------
# 1. BINGKAI ATAS
# ------------------------------------------------------------------------------
print("#", end="")
for k_1234 in range(4 * n_1234 + 5):
    print("=", end="")
print("#")

# ------------------------------------------------------------------------------
# 2. FASE 1: JAM PASIR ATAS (Mundur dari N ke 1)
# ------------------------------------------------------------------------------
for baris_1234 in range(n_1234, 0, -1):
    # Sisi kiri: Bingkai dan padding
    print("| ", end="")
    
    # Spasi penyeimbang kiri
    for s_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")
        
    # Deret angka mundur (baris -> 1)
    for a_1234 in range(baris_1234, 0, -1):
        print(a_1234, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju (1 -> baris)
    for a_1234 in range(1, baris_1234 + 1):
        print("", a_1234, end="")
        
    # Spasi penyeimbang kanan
    for s_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")
        
    # Sisi kanan: Padding dan bingkai
    print(" |")

# ------------------------------------------------------------------------------
# 3. FASE 2: POROS TITIK PUSAT (Nol / Singularity)
# ------------------------------------------------------------------------------
print("|", end="")
for s_1234 in range(2 * n_1234 + 1):
    print(" ", end="")

print("<*>", end="")

for s_1234 in range(2 * n_1234 + 1):
    print(" ", end="")
print("|")

# ------------------------------------------------------------------------------
# 4. FASE 3: JAM PASIR BAWAH (Maju dari 1 ke N)
# ------------------------------------------------------------------------------
for baris_1234 in range(1, n_1234 + 1):
    # Sisi kiri: Bingkai dan padding
    print("| ", end="")
    
    # Spasi penyeimbang kiri
    for s_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")
        
    # Deret angka mundur (baris -> 1)
    for a_1234 in range(baris_1234, 0, -1):
        print(a_1234, end=" ")
        
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju (1 -> baris)
    for a_1234 in range(1, baris_1234 + 1):
        print("", a_1234, end="")
        
    # Spasi penyeimbang kanan
    for s_1234 in range(2 * (n_1234 - baris_1234)):
        print(" ", end="")
        
    # Sisi kanan: Padding dan bingkai
    print(" |")

# ------------------------------------------------------------------------------
# 5. BINGKAI BAWAH
# ------------------------------------------------------------------------------
print("#", end="")
for k_1234 in range(4 * n_1234 + 5):
    print("=", end="")
print("#")
```

**Penjelasan :**
1. Bingkai Pembatas: Menggunakan loop `k_2030` untuk mencetak karakter `=` sebanyak `4 * N + 5` di antara dua ujung `#`.
2. Fase 1 (Jam Pasir Atas): Menggunakan loop `baris_2030` berulang mundur dari `N` hingga `1`. Pada setiap baris, loop `s_2030` mencetak spasi penyeimbang, loop `a_2030` mencetak deret mundur, lalu disisipkan poros `<*>`, dilanjutkan deret angka maju dan spasi penyeimbang kanan.
3. Fase 2 (Poros Pusat): Mencetak titik singularity berupa karakter `<*>` tepat di tengah yang diapit oleh `2 * N + 1` spasi di kiri dan kanan di dalam pembatas `|`.
4. Fase 3 (Jam Pasir Bawah): Menggunakan loop `baris_2030` berulang maju dari `1` hingga `N` dengan struktur perulangan simetris untuk membentuk alas jam pasir.
5. Aturan Kepatuhan: Seluruh variabel menggunakan suffix `_2030` dan pembentukan string murni menggunakan nested loop tanpa perkalian operator string."
