## 1.1 Latar Belakang
Di era transformasi digital saat ini, efisiensi dalam pengembangan perangkat lunak menjadi kunci utama keberhasilan sebuah proyek teknologi. Bagi seorang programer, efisiensi tidak hanya ditentukan oleh kemampuan menulis kode secara logis, tetapi juga oleh pemilihan ekosistem kerja yang tepat. Ekosistem tersebut mencakup bahasa pemrograman yang digunakan, lingkungan pengembangan (environment), serta sistem manajemen kode. Ketiga aspek ini terangkum dalam pilar utama praktikum ini, yaitu Python, Visual Studio Code (VS Code), dan GitHub.

Mengapa konsep-konsep ini penting dipelajari?
Pertama, Python telah bergeser dari sekadar bahasa pemrograman alternatif menjadi standar industri dalam berbagai bidang, termasuk kecerdasan buatan (AI), analisis data, hingga pengelolaan struktur data kompleks. Mempelajari Python sangat penting karena karakteristiknya yang mengutamakan keterbacaan kode (readability). Hal ini memungkinkan programer pemula maupun profesional untuk lebih fokus pada penyelesaian masalah logis (problem solving) daripada terjebak dalam kerumitan sintaksis.

Kedua, proses penulisan kode memerlukan wadah yang adaptif. VS Code menjadi penting untuk dikuasai karena statusnya sebagai salah satu code editor paling populer di industri modern. Memahami cara mengonfigurasi VS Code sejak dini melatih mahasiswa untuk membangun lingkungan kerja (development environment) yang efisien, mempercepat proses penulisan, serta mempermudah pencarian kesalahan (debugging) dalam kode program.

Ketiga, dalam dunia profesional, seorang programer tidak pernah bekerja dalam isolasi. Konsep Version Control System (VCS) melalui Git dan GitHub menjadi kompetensi wajib. Mempelajari GitHub sangat penting karena mengajarkan cara mendokumentasikan setiap perubahan kode secara terstruktur (history tracking). Tanpa pemahaman ini, risiko kehilangan data atau benturan kode (code collision) saat bekerja dalam tim akan sangat tinggi.

**Penjelasan :**
1. Komentar Program, baris tersebut merupakan komentar pada Python. Komentar diawali dengan tanda **#** dan tidak akan dijalankan sebagai perintah oleh program. Komentar digunakan untuk memberikan keterangan atau penjelasan mengenai kode agar lebih mudah dipahami.
2. Multi-line String, bagian tersebut menggunakan tiga tanda kutip **(""")** untuk membuat multi-line string, yaitu teks yang dapat ditulis dalam beberapa baris. Dalam konteks program ini, bagian tersebut digunakan sebagai keterangan atau dokumentasi sederhana yang tidak menghasilkan output karena tidak disimpan ke dalam variabel maupun ditampilkan dengan **print()**.
3. Menampilkan Teks, fungsi print() digunakan untuk menampilkan data atau teks ke layar/output. Pada kode tersebut, teks "Hello world" akan ditampilkan pada hasil eksekusi program.
4. Membuat Baris Kosong, fungsi print() tanpa isi digunakan untuk menghasilkan satu baris kosong pada output. Hal ini berguna untuk memberikan jarak antara satu output dengan output lainnya sehingga hasil program lebih mudah dibaca.
5. Menampilkan Kalimat, kode tersebut menggunakan fungsi print() untuk menampilkan kalimat “Program ini menghasilkan” ke layar.
6. Menampilkan Kalimat Berikutnya, kode ini digunakan untuk menampilkan teks “Empat baris output”. Fungsi print() secara default akan berpindah ke baris berikutnya setelah menampilkan teks.
7. Menampilkan Karakter Khusus, kode tersebut digunakan untuk menampilkan karakter / dan \ pada output. Karakter \ disebut backslash dan memiliki fungsi khusus dalam Python. Oleh karena itu, untuk menampilkan satu karakter backslash secara langsung, diperlukan penulisan \\.
