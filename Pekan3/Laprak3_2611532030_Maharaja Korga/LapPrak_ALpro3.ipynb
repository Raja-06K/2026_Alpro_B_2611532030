## 2.1 Dasar Teori

### 2.1.1 Bahasa Pemrograman Python
Python merupakan bahasa pemrograman tingkat tinggi yang bersifat interpreted dan berfokus pada keterbacaan kode. Fleksibilitas Python memudahkan manipulasi variabel dan ekspresi logika secara ringkas.

### 2.1.2 Operator Aritmatika
Operator aritmatika digunakan untuk melakukan operasi matematika dasar seperti penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat/floor division (//), sisa bagi/modulo (%), dan perpangkatan (**).

### 2.1.3 Operator Assignment
Operator assignment digunakan untuk memberikan atau memperbarui nilai pada suatu variabel. Selain penugasan biasa (=), Python mendukung gabungan penugasan dengan operasi aritmatika (seperti +=, -=, *=, /=, //=, %=, dan **=).

### 2.1.4 Operator Bitwise
Operator bitwise beroperasi langsung pada bit-bit biner dari suatu nilai integer. Operator ini meliputi AND (&), OR (|), XOR (^), NOT (~), Shift Left (<<), dan Shift Right (>>).

### 2.1.5 Operator Keanggotaan dan Identitas
- Keanggotaan (in, not in): Digunakan untuk memeriksa apakah suatu nilai terdapat di dalam suatu urutan/koleksi data (misalnya list atau string).
- Identitas (is, is not): Digunakan untuk memeriksa apakah dua variabel merujuk pada objek/lokasi memori yang sama.

### 2.1.6 Operator Logika
Operator logika mengevaluasi ekspresi bernilai boolean (True / False). Terdiri dari konjungsi (and), disjungsi (or), dan negasi (not).

### 2.1.7 Operator Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan nilai boolean (True atau False). Terdiri dari lebih besar (>), lebih kecil (<), lebih besar/sama dengan (>=), lebih kecil/sama dengan (<=), sama dengan (==), dan tidak sama dengan (!=).

## 2.2 Kode Program

### 2.2.1 Program Aritmatika "Aritmatika_2611532030.py"

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2030 = int(input("input angka-1: "))
angka2_2030 = int(input("input angka-2: "))

# Penjumlahan
hasil_2030 = angka1_2030 + angka2_2030
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2030)

# Pengurangan
hasil_2030 = angka1_2030 - angka2_2030
print("\nOperator Pengurangan")
print("Hasil =", hasil_2030)

# Perkalian
hasil_2030 = angka1_2030 * angka2_2030
print("\nOperator Perkalian")
print("Hasil =", hasil_2030)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2030 != 0:
    hasil_2030 = angka1_2030 / angka2_2030
    print("\nOperator Pembagian")
    print("Hasil =", hasil_2030)

    hasil_2030 = angka1_2030 // angka2_2030
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2030)

    hasil_2030 = angka1_2030 % angka2_2030
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_2030)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2030 = angka1_2030 ** angka2_2030
print("\nOperator Pangkat")
print("Hasil =", hasil_2030)
```

**Penjelasan Kode Program Aritmatika:**
1. Input & Casting: Mengambil masukan dua buah angka dari pengguna menggunakan input() dan mengonversinya menjadi tipe integer (int).
2. Operasi Aritmatika Dasar: Melakukan kalkulasi penjumlahan (+), pengurangan (-), dan perkalian (*).
3. Validasi Pembagian (Pengecekan Pembagi != 0): Menggunakan kondisional if angka2_2030 != 0 untuk mencegah ZeroDivisionError saat melakukan pembagian riil (/), pembagian bulat (//), dan sisa bagi (%).
4. Perpangkatan: Menghitung angka1_2030 pangkat angka2_2030 menggunakan operator **.

### 2.2.2 Program Assignment "Assignment_2611532030.py"

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menggunakan fungsi input()
# Nilai program yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_2030 = int(input("input angka-1: "))
angka2_2030 = int(input("input angka-2: "))

print("\nNilai awal angka1 =", angka1_2030)
print("\nNilai awal angka2 =", angka2_2030)

# assignment biasa
hasil = angka1_2030
print("\nOperator assignment biasa (=)")
print("hasil =", hasil)

# assignment penambahan
hasil = angka1_2030
hasil += angka2_2030
print("\nOperator assignment penambahan (+=)")
print("hasil =", hasil)

# assignment pengurangan
hasil = angka1_2030
hasil -= angka2_2030
print("\nOperator assignment pengurangan (-=)")
print("hasil =", hasil)

# assignment perkalian
hasil = angka1_2030
hasil *= angka2_2030
print("\nOperator assignment perkalian (*=)")
print("hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2030 != 0:
    hasil = angka1_2030
    hasil /= angka2_2030
    print("\nAssignment pembagian (/=)")
    print("hasil =", hasil)
    # Operator tambahan
    hasil = angka1_2030
    hasil //= angka2_2030
    print("\nAssignment pembagian bulat (//=)")
    print("hasil =", hasil)
    hasil = angka1_2030
    hasil %= angka2_2030
    print("\nAssignment sisa bagi (%=)")
    print("hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment berpangkat
hasil = angka1_2030
hasil **= angka2_2030
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil)
```

**Penjelasan Kode Program Assignment:**
1. Assignment Biasa (=): Menugaskan nilai awal angka1_2030 langsung ke variabel hasil.
2. Assignment Gabungan (Penambahan, Pengurangan, Perkalian): Secara berturut-turut memperbarui nilai variabel hasil dengan menambahkan (+=), mengurangkan (-=), dan mengalikan (*=) dengan nilai angka2_2030.
3. Assignment Pembagian & Modulo: Memperbarui variabel dengan nilai hasil pembagian (/=), pembagian bulat (//=), serta modulo (%=) setelah memastikan angka2_2030 bukan nol.
4. Assignment Perpangkatan: Memperbarui nilai hasil dengan dipangkatkan angka2_2030 (**=).

### 2.2.3 Program Bitwise "Bitwise_2611532030.py"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n===================================")
print("3. OPERATOR BITWISE")
print("===================================")

angka1_2030 = int(input("Masukkan angka bitwise-1: "))
angka2_2030 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2030, "| biner=", bin(angka1_2030))
print("angka2 =", angka2_2030, "| biner=", bin(angka2_2030))

# Bitwise AND
hasil_2030 = angka1_2030 & angka2_2030
print("\nBitwise AND (&)")
print(angka1_2030, "&", angka2_2030, "=", hasil_2030)
print("Biner hasil =", bin(hasil_2030))
print("Biner hasil (8 bit) =", format(hasil_2030, '08b'))

# Bitwise OR
hasil_2030 = angka1_2030 | angka2_2030
print("\nBitwise OR (|)")
print(angka1_2030, "|", angka2_2030, "=", hasil_2030)
print("Biner hasil =", bin(hasil_2030))
print("Biner hasil (8 bit) =", format(hasil_2030, '08b'))

# Bitwise XOR
hasil_2030 = angka1_2030 ^ angka2_2030
print("\nBitwise XOR (^)")
print(angka1_2030, "^", angka2_2030, "=", hasil_2030)
print("Biner hasil =", bin(hasil_2030))
print("Biner hasil (8 bit) =", format(hasil_2030, '08b'))

# Bitwise NOT
hasil_2030 = ~angka1_2030
print("\nBitwise NOT (~)")
print("~", angka1_2030, "=", hasil_2030)
print("Biner hasil =", bin(hasil_2030))
print("Biner hasil (8 bit) =", format(hasil_2030, '08b'))

# Bitwise geser kiri
jumlah_geser_2030 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2030 = angka1_2030 << jumlah_geser_2030
print("\nBitwise geser kiri (<<)")
print(angka1_2030, "<<", jumlah_geser_2030, "=", hasil_2030)
print("Biner hasil =", bin(hasil_2030))
print("Biner hasil (8 bit) =", format(hasil_2030, '08b'))

# Bitwise geser kanan
jumlah_geser_2030 = int(input("\nMasukkan jumlah pergeseran bit: "))
hasil_2030 = angka1_2030 >> jumlah_geser_2030
print("\nBitwise geser kanan (>>)")
print(angka1_2030, ">>", jumlah_geser_2030, "=", hasil_2030)
print("Biner hasil =", bin(hasil_2030))
print("Biner hasil (8 bit) =", format(hasil_2030, '08b'))
```

**Penjelasan Kode Program Bitwise:**
1. Konversi ke Biner: Menggunakan fungsi bawaan bin() dan format(..., '08b') untuk memperlihatkan nilai dalam representasi bit biner.
2. Logika Biner (&, |, ^, ~): Menjalankan operasi biner bit-demi-bit (AND, OR, XOR, dan Komplemen NOT dua).
3. Pergeseran Bit (<<, >>): Menggeser posisi bit angka ke kiri (<<) atau ke kanan (>>) sebanyak nilai jumlah_geser_2030.

### 2.2.4 Program Lainnya "Lainnya_2611532030.py"

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_2030 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_2030 = [int(angka.strip()) for angka in input_data_2030.split(",")]

nilai_dicari_2030 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_2030 = nilai_dicari_2030 in data_2030
print("\nOperator keanggotaan IN")
print(nilai_dicari_2030, "in", data_2030, "=", hasil_2030)

# Operator not in
hasil_2030 = nilai_dicari_2030 not in data_2030
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_2030, "not in", data_2030, "=", hasil_2030)


print("\n==================================")
print("2. OPERATOR IDENTITAS")
print("\n==================================")

# objek1 menggunakan list dari input pengguna
objek1_2030 = data_2030

# objek2 merujuk pada objek yang sama dengan objek1
objek2_2030 = objek1_2030

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_2030 = data_2030.copy()

# Operator is
hasil_2030 = objek1_2030 is objek2_2030
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_2030)

# Operator is not
hasil_2030 = objek1_2030 is not objek3_2030
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_2030)

# Membandingkan identitas dan nilai
print("\nMembandingkan identitas dan nilai:")
print("objek1 is objek2 =", objek1_2030 is objek3_2030)
print("objek1 == objek3 =", objek1_2030 == objek3_2030)
```

**Penjelasan Kode Program Lainnya:**
1. Pemrosesan List: Memecah string input berpemisah koma dan mengonversinya menjadi list integer.
2. Keanggotaan (in, not in): Memeriksa keberadaan nilai nilai_dicari_2030 di dalam list data_2030.
3. Identitas (is, is not):
- objek2_2030 = objek1_2030 merujuk ke lokasi memori yang sama sehingga is menghasilkan True.
- objek3_2030 = data_2030.copy() menyalin nilai ke lokasi memori baru sehingga objek1 is objek3 bernilai False, namun nilai persamaannya (==) bernilai True.

### 2.2.5 Program Logika "Logika_2611532030.py"

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2030 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2030 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_2030)
print("\nA2 =", a2_2030)

# Konjungsi: bernilai True jika keduanya true
hasil_2030 = a1_2030 and a2_2030
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_2030)

# Disjungsi: bernilai True jika salah satunya true 
hasil_2030 = a1_2030 or a2_2030
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil_2030)

# Negasi A1: membalik nilai A1
hasil_2030 = not a1_2030
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_2030)

# Negasi A2: membalik nilai A2
hasil_2030 = not a2_2030
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil_2030)

# XOR: bernilai True jika kedua nilai berbeda
hasil_2030 = a1_2030 != a2_2030
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_2030)
```

**Penjelasan Kode Program Logika:**
1. Casting String ke Boolean: Mengubah string masukan pengguna menjadi boolean True jika bernilai "true" (mengabaikan kapitalisasi dan spasi).
2. Evaluasi Logika (and, or, not): Menguji hukum gerbang logika konjungsi, disjungsi, serta negasi pada variabel masukan.
3. Disjungsi Eksklusif (XOR): Menggunakan operator pertidaksamaan (!=) untuk merepresentasikan kondisi XOR (bernilai True jika hanya salah satu operand bernilai True).

### 2.2.6 Program Perbandingan "Perbandingan_2611532030.py"

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit terakhir nim contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_2030 = int(input("input angka-1: "))
angka2_2030 = int(input("input angka-2: "))

# Lebih besar dari
hasil_2030 = angka1_2030 > angka2_2030
print("\nOperator Lebih Besar Dari")
print("angka1_2030 > angka2_2030 =", hasil_2030)

# Lebih kecil dari
hasil_2030 = angka1_2030 < angka2_2030
print("\nOperator Lebih Kecil Dari")
print("angka1_2030 < angka2_2030 =", hasil_2030)

# Lebih besar dari atau sama dengan
hasil_2030 = angka1_2030 >= angka2_2030
print("\nOperator Lebih Besar Dari atau Sama Dengan")
print("angka1_2030 >= angka2_2030 =", hasil_2030)

# Lebih kecil dari atau sama dengan
hasil_2030 = angka1_2030 <= angka2_2030
print("\nOperator Lebih Kecil Dari atau Sama Dengan")
print("angka1_2030 <= angka2_2030 =", hasil_2030)

# Sama dengan
hasil_2030 = angka1_2030 == angka2_2030
print("\nOperator Sama Dengan")
print("angka1_2030 == angka2_2030 =", hasil_2030)

# Tidak sama dengan
hasil_2030 = angka1_2030 != angka2_2030 
print("\nOperator Tidak Sama Dengan")
print("angka1_2030 != angka2_2030 =", hasil_2030)

# Tambahan: perbandingan berantai dalam Python
hasil_2030 = 0 < angka1_2030 < 100
print("\nOperator Perbandingan Berantai")
print("0 < angka1_2030 < 100 =", hasil_2030)

hasil_2030 = 0 < angka2_2030 < 100
print("0 < angka2_2030 < 100 =", hasil_2030)
```

**Penjelasan Kode Program Perbandingan:**
1. Evaluasi Relasional Dasar: Menguji hubungan numerik dua angka menggunakan operator >, <, >=, <=, ==, dan !=. Hasil evaluasi selalu bertipe bool (True atau False).
2. Perbandingan Berantai (Chained Comparison): Menguji rentang nilai (contoh: 0 < angka1_2030 < 100) secara langsung tanpa perlu penggabungan ekplisit dua pernyataan dengan and.

a. Kode Program "tugas3_2611532030"

```python
# tugas3_2611532030.py
# Sistem Simulasi Transaksi dan Validasi Akses Toko

print("=== SISTEM TRANSAKSI TOKO ===\n")

# 1. INPUT DATA PELANGGAN & TRANSAKSI
nama_2030 = input("Masukkan Nama Pelanggan : ")
status_2030 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_2030 = float(input("Masukkan Total Belanja : "))
jumlah_2030 = int(input("Masukkan Jumlah Barang : "))
promo_2030 = input("Masukkan Kode Promo : ")

# Daftar kode promo yang tersedia di toko
daftar_promo_2030 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

# 2. OPERATOR PERBANDINGAN
syarat_belanja_2030 = total_2030 >= 200000
syarat_barang_2030 = jumlah_2030 >= 3
is_member_2030 = status_2030.lower() == "member"

# 3. OPERATOR LOGIKA
diskon_member_2030 = is_member_2030 and syarat_belanja_2030
dapat_promo_2030 = (is_member_2030 and syarat_belanja_2030) or syarat_barang_2030
bukan_member_2030 = not is_member_2030

# 4. OPERATOR KEANGGOTAAN (in / not in)
promo_tersedia_2030 = promo_2030 in daftar_promo_2030
promo_tidak_tersedia_2030 = promo_2030 not in daftar_promo_2030

# 5. OPERATOR ARITMATIKA
if diskon_member_2030 or dapat_promo_2030:
    diskon_2030 = total_2030 * 0.1
else:
    diskon_2030 = 0.0

bayar_2030 = total_2030 - diskon_2030
rata_2030 = bayar_2030 / jumlah_2030
sisa_barang_2030 = jumlah_2030 % 2

# 6. OPERATOR PENUGASAN (assignment)
poin_2030 = 0
poin_2030 += jumlah_2030
if dapat_promo_2030:
    poin_2030 *= 2
bayar_2030 -= 0

# 7. OPERATOR IDENTITAS (is / is not)
objek1_2030 = ["status", is_member_2030]
objek2_2030 = ["status", is_member_2030]
objek3_2030 = objek1_2030
identitas_sama_2030 = objek1_2030 is objek3_2030
identitas_beda_2030 = objek1_2030 is not objek2_2030
nilai_sama_2030 = objek1_2030 == objek2_2030

# 8. OPERATOR BITWISE
bit_member_2030   = 0b0001 if is_member_2030 else 0b0000
bit_belanja_2030  = 0b0010 if syarat_belanja_2030 else 0b0000
bit_barang_2030   = 0b0100 if syarat_barang_2030 else 0b0000
bit_promo_2030    = 0b1000 if promo_tersedia_2030 else 0b0000

# OR (|) -> menggabungkan seluruh kondisi menjadi satu kode status
kode_status_2030 = bit_member_2030 | bit_belanja_2030 | bit_barang_2030 | bit_promo_2030

# AND (&) -> memeriksa apakah bit tertentu aktif
cek_member_2030 = kode_status_2030 & 0b0001
cek_promo_2030  = kode_status_2030 & 0b1000

# XOR (^) -> membandingkan perbedaan status dengan kode referensi
kode_referensi_2030 = 0b1011
selisih_status_2030 = kode_status_2030 ^ kode_referensi_2030

# Shift kiri (<<) -> contoh tambahan operator bitwise
geser_status_2030 = kode_status_2030 << 1

# Hak akses berdasarkan hasil bitwise
member_access_2030 = bool(cek_member_2030)
promo_access_2030 = bool(cek_promo_2030)
free_shipping_access_2030 = (promo_2030 == "GRATISONGKIR") and promo_tersedia_2030

# OUTPUT PROGRAM
print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan       : {nama_2030}")
print(f"Status Pelanggan     : {status_2030}")
print(f"Total Belanja        : Rp{total_2030:.0f}")
print(f"Jumlah Barang        : {jumlah_2030}")
print(f"Kode Promo           : {promo_2030}")

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {syarat_belanja_2030}")
print(f"Jumlah Barang >= 3         : {syarat_barang_2030}")
print(f"Status Member              : {is_member_2030}")
print(f"Kode Promo Tersedia        : {promo_tersedia_2030}")
print(f"Mendapatkan Diskon Member  : {diskon_member_2030}")
print(f"Mendapatkan Promo          : {dapat_promo_2030}")

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                     : Rp{diskon_2030:.0f}")
print(f"Total Pembayaran           : Rp{bayar_2030:.0f}")
print(f"Rata-rata Harga Barang     : Rp{rata_2030:.0f}")
print(f"Sisa Bagi Jumlah Barang %2 : {sisa_barang_2030}")
print(f"Poin Pelanggan             : {poin_2030}")

print("\n=== OPERATOR IDENTITAS ===")
print(f"objek1 is objek3 (sama)     : {identitas_sama_2030}")
print(f"objek1 is not objek2 (beda) : {identitas_beda_2030}")
print(f"objek1 == objek2 (nilai)    : {nilai_sama_2030}")

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Status (biner)  : {bin(kode_status_2030)}")
print(f"Kode Status (desimal): {kode_status_2030}")
print(f"Member Access        : {member_access_2030}")
print(f"Promo Access         : {promo_access_2030}")
print(f"Free Shipping Access : {free_shipping_access_2030}")

print("\n=== OPERASI BITWISE ===")
print(f"Cek Member  -> {bin(kode_status_2030)} & 0001 = {bin(cek_member_2030)} ({cek_member_2030})")
print(f"Cek Promo   -> {bin(kode_status_2030)} & 1000 = {bin(cek_promo_2030)} ({cek_promo_2030})")
print(f"XOR Ref     -> {bin(kode_status_2030)} ^ {bin(kode_referensi_2030)} = {bin(selisih_status_2030)} ({selisih_status_2030})")
print(f"Shift Kiri  -> {bin(kode_status_2030)} << 1 = {bin(geser_status_2030)} ({geser_status_2030})")

print("\n=== SELESAI ===")
```

**Penjelasan :**
1. Input Data Pelanggan & Transaksi
- Mengambil input nama pelanggan, status keanggotaan (member/nonmember), total belanja, jumlah barang, dan kode promo dari pengguna menggunakan fungsi input().
- Tipe data total belanja dikonversi ke float untuk memungkinkan nilai pecahan, sedangkan jumlah barang dikonversi ke int.
- Didefinisikan pula sebuah list daftar_promo_2030 yang berisi daftar kode promo resmi toko ("HEMAT10", "HEMAT20", "GRATISONGKIR").

2. Operator Perbandingan
- syarat_belanja_2030: Menggunakan operator >= untuk mengevaluasi apakah total belanja pelanggan mencapai batas minimal Rp200.000.   
- syarat_barang_2030: Menggunakan operator >= untuk mengevaluasi apakah jumlah barang yang dibeli minimal 3 buah.   
- is_member_2030: Menggunakan operator kesamaan == untuk memeriksa apakah status pelanggan yang diinputkan bernilai "member" (setelah diubah ke huruf kecil menggunakan .lower()).   

3. Operator Logika
- diskon_member_2030: Menggunakan operator and yang bernilai True hanya jika pelanggan merupakan member dan memenuhi syarat belanja minimal Rp200.000.   
- dapat_promo_2030: Menggunakan kombinasi and dan or untuk menentukan kelayakan promo (bernilai True jika pelanggan adalah member yang belanja $\ge$ Rp200.000 atau membeli $\ge$ 3 barang).   
- bukan_member_2030: Menggunakan operator negasi not untuk membalikkan nilai boolean dari is_member_2030.   

4. Operator Keanggotaan & Identitas/Lainnya
- promo_tersedia_2030: Menggunakan operator in untuk memeriksa apakah kode promo yang diinputkan pelanggan terdaftar dalam daftar_promo_2030.
- promo_tidak_tersedia_2030: Menggunakan operator not in untuk mengevaluasi kondisi jika kode promo tidak terdaftar dalam sistem toko.
- objek1_2030 dan objek2_2030 dibuat sebagai dua list terpisah dengan isi/nilai yang sama persis.
- objek3_2030 = objek1_2030: Membuat objek3_2030 merujuk ke lokasi memori yang sama dengan objek1_2030.
- objek1_2030 is objek3_2030: Bernilai True karena merujuk ke objek/alamat memori yang sama.
- objek1_2030 is not objek2_2030: Bernilai True karena meskipun nilainya sama, keduanya merupakan objek yang teralokasi di memori berbeda.
- objek1_2030 == objek2_2030: Bernilai True karena operator == hanya membandingkan nilai kontennya, bukan lokasi memorinya.

5. Operator Aritmatika
- Menghitung nilai diskon sebesar 10% (total_2030 * 0.1) jika pelanggan berhak mendapatkan diskon/promo, atau 0.0 jika tidak.   
- Menghitung bayar_2030 dengan mengurangkan total belanja dengan diskon (total_2030 - diskon_2030).   
- Menghitung rata-rata harga barang menggunakan operator pembagian / (bayar_2030 / jumlah_2030).   
- Menghitung sisa bagi jumlah barang terhadap angka 2 menggunakan operator modulo % (jumlah_2030 % 2).   

6. Operator Assignment
- poin_2030 += jumlah_2030: Menambahkan jumlah barang yang dibeli ke dalam variabel poin_2030.   
- poin_2030 *= 2: Mengalikan poin pelanggan menjadi 2 kali lipat menggunakan operator *=, jika pelanggan memenuhi syarat promo.   
- bayar_2030 -= 0: Contoh operasi assignment pengurangan pada total pembayaran.   

7. Operator Bitwise & Penentuan Hak Akses
- Menyusun bit-bit representasi kondisi pelanggan menggunakan format biner:
    - Bit 0 (0b0001): Status Member.
    - Bit 1 (0b0010): Syarat Belanja ($\ge$ Rp200.000).
    - Bit 2 (0b0100): Syarat Barang ($\ge$ 3).
    - Bit 3 (0b1000): Promo Tersedia.
- Bitwise OR (|): Menggabungkan keempat flag bit tersebut menjadi satu nilai biner utuh (kode_status_2030).
- Bitwise AND (&): Digunakan untuk mengekstrak/memeriksa apakah bit tertentu aktif (misalnya kode_status_2030 & 0b0001 untuk mengecek bit member).   
- Bitwise XOR (^): Membandingkan perbedaan bit antara kode_status_2030 dengan kode_referensi_2030 (0b1011).   
- Bitwise Shift Left (<<): Menggeser posisi bit kode_status_2030 ke kiri sebanyak 1 bit.   
- Konversi nilai ekstraksi bit ke bool() digunakan untuk memberikan status hak akses pelanggan (member_access_2030, promo_access_2030, dan free_shipping_access_2030).
